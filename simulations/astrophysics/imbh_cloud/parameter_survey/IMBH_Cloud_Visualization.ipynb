# IMBH-Cloud Tidal Disruption Visualization

This notebook serves the SPH simulation visualization from Google Drive.

## Setup Instructions
1. Upload your `parameter_survey` folder to Google Drive
2. Run all cells below
3. Click the generated URL to view the visualization

````
# Step 1: Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')
````

````
# Step 2: Configure paths
# EDIT THIS PATH to match your Google Drive folder location
DRIVE_PATH = "/content/drive/MyDrive/sph-code-simulation/simulations/astrophysics/imbh_cloud/parameter_survey"

import os
if os.path.exists(DRIVE_PATH):
    print(f"Found folder: {DRIVE_PATH}")
    print(f"Contents: {os.listdir(DRIVE_PATH)[:10]}...")
else:
    print(f"ERROR: Folder not found: {DRIVE_PATH}")
    print("Please edit DRIVE_PATH above to match your Google Drive location")
````

````
# Step 3: Start HTTP server
import subprocess
import time

# Kill any existing server
!pkill -f "python.*http.server" 2>/dev/null || true

# Change to the visualization directory
os.chdir(DRIVE_PATH)
print(f"Serving from: {os.getcwd()}")

# Start server in background
server_process = subprocess.Popen(
    ['python', '-m', 'http.server', '8080'],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE
)
time.sleep(2)
print("Server started on port 8080")
````

````
# Step 4: Get public URL
from google.colab.output import eval_js

# Get the proxy URL for port 8080
proxy_url = eval_js("google.colab.kernel.proxyPort(8080)")
viz_url = f"{proxy_url}viz/imbh_cloud_viz.html"

print("="*60)
print("VISUALIZATION URL (click to open):")
print("="*60)
print(f"\n{viz_url}\n")
print("="*60)
print("\nNote: This URL is temporary and only works while this notebook is running.")
print("Share the Colab notebook link for others to run their own instance.")
````

````
# Optional: Display visualization inline (may have limited interactivity)
from IPython.display import IFrame
IFrame(viz_url, width=1200, height=800)
````

````
# Keep the server running
# Run this cell to keep the notebook alive
import time
print("Server is running. Keep this cell running to maintain the visualization.")
print("Press the stop button to terminate.")

try:
    while True:
        time.sleep(60)
        print(".", end="", flush=True)
except KeyboardInterrupt:
    print("\nServer stopped.")
````

## Troubleshooting

### "Folder not found" error
- Check that your folder is uploaded to Google Drive
- Edit `DRIVE_PATH` in Step 2 to match your folder location
- Use the file browser on the left to navigate and find the correct path

### Visualization not loading
- Make sure the folder structure is:
  ```
  parameter_survey/
  ├── viz/
  │   ├── imbh_cloud_viz.html
  │   ├── css/
  │   ├── js/
  │   └── datasets.json
  ├── parabolic_rp0.4pc/
  │   ├── adiabatic/results/
  │   └── cooling/results/
  └── ...
  ```

### Slow loading
- The simulation data files are large (~11MB each)
- Initial loading may take 1-2 minutes
- Consider using fewer snapshots for faster loading
